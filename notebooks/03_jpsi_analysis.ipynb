# J/ψ Mass Analysis

This notebook analyzes opposite-charge muon pairs from the CMS dataset and reconstructs their invariant mass distribution to search for the J/ψ particle.

```python
import uproot
import awkward as ak
import numpy as np
import matplotlib.pyplot as plt 

all_masses = np.load("../data/processed/dimuon_masses.npy")

print("Total invariant masses:", len(all_masses))
print(all_masses[:10])

plt.figure(figsize=(10,6))

plt.hist(
    all_masses,
    bins=30,
    range=(2.9, 3.3)
)

plt.xlabel("Invariant Mass (GeV)")
plt.ylabel("Number of muon pairs")
plt.title("Invariant Mass of Opposite-Charge Muon Pairs")

plt.axvline(3.1, linestyle="--", label="Expected J/ψ mass ≈ 3.1 GeV")

plt.legend()
plt.show()

jpsi_candidates = all_masses[
    (all_masses > 2.9) & (all_masses < 3.3)
]

print("Number of candidates in J/ψ region:", len(jpsi_candidates))
print(jpsi_candidates[:20])

np.save("jpsi_candidates.npy", jpsi_candidates)
print("Saved", len(jpsi_candidates), "J/ψ-region candidates")
```
